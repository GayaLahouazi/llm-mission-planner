# LLM + Z3 mission planner: benchmark on a free Colab GPU

Runtime > Change runtime type > **T4 GPU**, then Run all (about 25 min for 30 missions).

The notebook clones the repo, loads an open LLM (Qwen2.5-3B-Instruct, Apache-2.0) with Hugging Face `transformers`, and runs `benchmark.py` in three settings: LLM alone, LLM + RAG, LLM + RAG + Z3 repair loop. Results are written to `results/benchmark_<model>.json`.

```
import json, glob
for f in glob.glob('results/benchmark_*.json'):
    r = json.load(open(f))
    print(f)
    for name, m in r['planners'].items():
        print(f"  {name:55s} valid {m['valid_rate']:.0%}  gap {m['mean_gap_to_optimal']}")
```
